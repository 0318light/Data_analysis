3.아래와 같은 구조를 갖는  데이터 프레임에서    (원본 객체 변경)  
1) '인문' 열을 추가 하여 영어와 국어 성적의 합계를 저장하고  
2) '과학' 열을 추가 하여 물리와 화학 성적의 합계를 저장하고  
3)  과학 성적으로 내림 차순으로 정렬하고  
4) print하는 code를 작성하시오.   


|이름|국어|영어|물리|화학|
|-|-|-|-|-|
|우석|90|85|70|50|
|인아|80|90|80|70|
|나라|75|85|80|90|
|민정|90|95|90|100|


```python
%reset -f
#과제 6
import pandas as pd
student_data = {'이름':['우석', '인아', '나라', '민정'],
                '국어':[90,80,75,90],
                '영어':[85,90,85,95],
                '물리':[70,80,80,90],
                '화학':[50,70,90,100] }
df=pd.DataFrame(student_data)
df.loc[:,'인문']=df.loc[:,'국어']+df.loc[:,'영어']
df.loc[:,'과학']=df.loc[:,'물리']+df.loc[:,'화학']
df.sort_values(by='과학',ascending=False,inplace=True)
print(df)
```

5.code에서 정의한 데이터에서   
1) 하나라도 NaN이 있는 행은 삭제하고 print하는 code  (df1에 저장 후 print)    
2) 모든 값이  NaN인  행을 삭제하고 print하는 code  (df2에 저장 후 print)   
3) NaN이 포함된 행을 삭제할때 데이터가 4개 이상 있어야 남기도록 하고  (df3)   print하는 code를 작성 하시오

```python
%reset -f
#과제 8
import pandas as pd

import numpy as np

raw_data = {'name': ['Jason', np.nan, 'Tina', 'Jake', 'Amy'],
            'age': [42, np.nan, 36, np.nan, 73],
            'math': [4, np.nan, np.nan, 2, 3],
            'physics': [25, np.nan, np.nan, 62, 70],
             'korean': [25, np.nan, np.nan, 62, 70]}

df = pd.DataFrame(raw_data)
print(df)
df1=df[:]
df1.dropna(inplace=True)
print('\n')
print(df1)
print('\n')


df2=df[:]
df2.dropna(how='all',inplace=True)
print(df2)
print('\n')


df3=df[:]
df3.dropna(thresh=4,inplace=True)
print(df3)
print('\n')
```

6.아래 code에서 정의하는 데이터 프레임에서    
1) 중복 행을 제거 하고 print하는 code(df1)  
2) c1,c3을 기준으로 중복행을 제거 하고 print하는 code(df2)  
를 작성 하시오.


```python
%reset -f
#과제 10
import pandas as pd

df=pd.DataFrame({'c1':['a','a','b','a','b'],
                 'c2':[2,1,2,1,2],
                 'c3':[1,1,2,2,2]})
print(df)
df1 = df[:]
df1.drop_duplicates(inplace=True)
print('\n')
print(df1)
print('\n')

df2 = df[:]
df2.drop_duplicates(subset=['c1','c3'],inplace=True)
print(df2)
```

5주차 실습

```python
import pandas as pd

data_list = {'이름':['우석','인아','나라','민정','서준'], '나이':[17, 14, 12, 19, 15], '구분':['고등','중등','초등','고등','중등'], '취미':['독서','산책','축구','독서','농구'], '학교':['선경','한국','우성','효성','상성']}
df = pd.DataFrame(data_list)
print(df)
print('\n')
# '이름' 열을 새로운 인덱스로 지정하고, df 객체에 변경사항 반영
df.set_index(['이름'], inplace=True)
print(df)
print('\n')

# '취미' 열 삭제
df.drop('취미', axis=1, inplace=True)
print(df)
print('\n')

# '나라' 행 삭제
df.drop('나라', axis=0, inplace=True)
print(df)
print('\n')

# '지민' 행 이름을 이용해 추가하여 '16, 중등, 효성' 데이터를 저장
df.loc['지민'] = [16, '중등', '효성']
print(df)
print('\n')

# column 이름을 '학교'에서 '교명'으로 변경
df.rename(columns={'학교':'교명'}, inplace=True)
print(df)
print('\n')

# 우석과 인아의 교명을 선경1, 한국1로 변경하고
df.loc['우석':'인아', '교명'] = ['선경1', '학교1']
print(df)
print('\n')

df.sort_values(by='나이',ascending=True,inplace=True)
print(df)
print('\n')

df.sort_values(by=['교명','나이'],ascending=[True,False],inplace=True)
print(df)
print('\n')

df['국적'] = '한국'
print(df)
```
